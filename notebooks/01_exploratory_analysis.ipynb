# Exploratory Analysis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

sns.set_theme(style="whitegrid")
```

```python
files = Path("../results").glob("*_exploration.csv")

df = pd.concat(
    [pd.read_csv(file) for file in files],
    ignore_index=True
)

df.head()
```

```python
# Dimensions du dataset
print("Nombre d'observations :", len(df))
print("Nombre de variables :", df.shape[1])
print("Nombre de robots :", df["robot"].nunique())
print("Valeurs manquantes :", df.isna().sum().sum())
print("Doublons :", df.duplicated().sum())

# Statistiques par robot
summary = df.groupby("robot").agg(
    observations=("step", "count"),
    duration_s=("time", "max"),
    distance_mean=("min_distance", "mean"),
    distance_median=("min_distance", "median"),
    distance_min=("min_distance", "min"),
    distance_max=("min_distance", "max"),
    distance_std=("min_distance", "std")
)

display(summary.round(3))

# Statistiques descriptives globales
display(df.describe().round(3))
```
